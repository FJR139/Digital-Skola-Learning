## Nomor 1

```python
import pandas as pd
df = pd.read_excel('Homework-DataFrame-Dataset.xlsx')
df_p = pd.read_excel('Homework-DataFrame-Dataset.xlsx', sheet_name='Paid-Transaction')
df_s = pd.read_excel('Homework-DataFrame-Dataset.xlsx', sheet_name='Seller')
df.head(10)
```

```python
full_name = (df_p['First Name'] + ' ' + df_p['Last Name']).str.title()
idx = df_p.columns.get_loc('First Name')
df_p.insert(idx + 1, 'Full Name', full_name)
df_p.head(10)
```

## nomor 2

```python
gmv = (df_p['Transaction Amount'] - df_p['Seller Discount']) + df_p['Delivery Fee']
```

```python
disc_idx = df_p.columns.get_loc('Seller Discount')
df_p.insert(disc_idx + 1, 'GMV', gmv)
df_p.head(10)
```

## nomor 3

```python
data_group = {
    'Home/ Food': 'Group 1',
    'Sports/ Hobi/ Otomotif': 'Group 1',
    'Fashion': 'Group 1',
    'Beauty/ Health': 'Group 1',
    'Gadget/ Komputer': 'Group 2',
    'Elektronik': 'Group 2',
    'Service/ Mokado': 'Group 2',
    'Babies/ Kids': 'Group 2'
}

df_p['group'] = df_p['Meta Category'].map(data_group)
df_p.head(10)
```

```python
df_p['Paid Date'] = pd.to_datetime(df_p['Paid Date'], format='%Y%m%d')
df_p['month'] = df_p['Paid Date'].dt.to_period('M')
df_p.head(10)
```

```python
df_pivot = df_p.pivot_table(index='month', columns='group', values='GMV', aggfunc='sum')
df_pivot.head(10)
```

## nomor 4

```python
df_merge = pd.merge(df_p, df_s, on='Order Number', how='left')
data_agustus = df_merge[df_merge['Paid Date'].dt.to_period('M') == '2017-08']
top_seller = data_agustus.groupby('Seller')['GMV'].sum().sort_values(ascending=False)
print(top_seller.head(10))
```

## nomor 5

```python
fashion_september = df_merge[(df_merge['month'] == '2017-09') & (df_merge['Meta Category'] == 'Fashion')]

top_seller_fashion = fashion_september.groupby('Seller')['GMV'].sum().sort_values(ascending=False)
print(top_seller_fashion.head(10))
```
