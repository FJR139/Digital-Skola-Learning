```python
def get_calculation(a, b):
  c = a+b
  d = a-b
  e = a*b
  f = a/b
  return c,d,e,f
```

```python
var1 = 20
var2 = 24

jumlah, kurang, kali, bagi=get_calculation(var1, var2)
```

```python
print("Penjumlahannya", jumlah)
print("Pengurangannya", kurang)
print("Perkaliannya", kali)
print("Pembagiannya", bagi)
```
